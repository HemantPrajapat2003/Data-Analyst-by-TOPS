```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
```

Task 1, default vs poster context

```python
spotify = pd.DataFrame({'day': days, 'songs': [24, 31, 28, 35, 42, 58, 49]})

sns.set()
plt.figure(figsize=(8, 5))
sns.barplot(data=spotify, x='day', y='songs')
plt.xlabel('Day'); plt.ylabel('Songs Streamed')
plt.title('Spotify Songs per Day (default context)')
plt.show()

sns.set(context='poster')
plt.figure(figsize=(8, 5))
sns.barplot(data=spotify, x='day', y='songs')
plt.xlabel('Day'); plt.ylabel('Songs Streamed')
plt.title('Spotify Songs per Day (poster context)')
plt.show()
```

Task 2, countplot with the colorblind palette

```python
sns.set()
cuisines = rng.choice(['Indian', 'Chinese', 'Italian', 'Mexican', 'Thai'], size=250, p=[0.35, 0.25, 0.2, 0.1, 0.1])
orders = pd.DataFrame({'cuisine': cuisines})

plt.figure(figsize=(8, 5))
sns.countplot(data=orders, x='cuisine', hue='cuisine', palette='colorblind', legend=False,
              order=orders['cuisine'].value_counts().index)
plt.xlabel('Cuisine Type')
plt.ylabel('Number of Orders')
plt.title('Food Orders by Cuisine (colorblind palette)')
plt.show()
```

Task 3, lineplot with the muted palette and notebook context

```python
sns.set(context='notebook', palette='muted')

steps = pd.DataFrame({
    'day': list(range(1, 15)) * 2,
    'steps': np.concatenate([rng.normal(8000, 1500, 14), rng.normal(6000, 1200, 14)]).round(),
    'person': ['Person A'] * 14 + ['Person B'] * 14,
})

plt.figure(figsize=(9, 5))
sns.lineplot(data=steps, x='day', y='steps', hue='person', marker='o')
plt.xlabel('Day')
plt.ylabel('Steps')
plt.title('Daily Step Counts Over Two Weeks')
plt.show()
```

Task 4, refactor from deep/talk to colorblind/poster

```python
data = pd.DataFrame({
    'category': ['A', 'B', 'C', 'D', 'E'] * 2,
    'group': ['Group 1'] * 5 + ['Group 2'] * 5,
    'value': rng.integers(20, 100, 10),
})

def make_plot(title):
    plt.figure(figsize=(9, 5))
    sns.barplot(data=data, x='category', y='value', hue='group')
    plt.title(title)
    plt.show()

sns.set(context='talk', palette='deep')
make_plot('Before: deep palette, talk context')

sns.set(context='poster', palette='colorblind')
make_plot('After: colorblind palette, poster context')
```

Task 5, Flipkart-style theme

```python
def flipkart_theme():
    sns.set(
        context='talk',
        palette='deep',
        style='whitegrid',
        font_scale=1.1,
        rc={'grid.linestyle': '--', 'axes.edgecolor': '#2874F0', 'axes.titleweight': 'bold'},
    )

flipkart_theme()

sales = pd.DataFrame({
    'category': ['Mobiles', 'Electronics', 'Fashion', 'Home', 'Grocery'],
    'sales': [520, 410, 380, 260, 190],
})

plt.figure(figsize=(10, 5))
sns.barplot(data=sales, x='category', y='sales', hue='category', legend=False)
plt.xlabel('Product Category')
plt.ylabel('Sales (₹ lakh)')
plt.title('Flipkart Sales by Category')
plt.show()
```
