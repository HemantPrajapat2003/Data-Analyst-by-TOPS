## Task 2: Weekly temperature trend with `lineplot()`

```python
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
temps = [32, 34, 33, 35, 36, 37, 34]

plt.figure(figsize=(8, 5))
sns.lineplot(x=days, y=temps, marker='o')
plt.xlabel('Day of the Week')
plt.ylabel('Temperature (°C)')
plt.title('Weekly Temperature Trend')
plt.show()
```

## Task 3: Axes-level vs figure-level

```python
# Axes-level: sns.lineplot()
fig, ax = plt.subplots(figsize=(8, 5))
sns.lineplot(x=days, y=temps, marker='o', ax=ax)
ax.set_xlabel('Day of the Week')
ax.set_ylabel('Temperature (°C)')
ax.set_title('Weekly Temperature Trend (axes-level: lineplot)')
plt.show()
```

```python
# Figure-level: sns.relplot(kind='line')
g = sns.relplot(x=days, y=temps, kind='line', marker='o', height=5, aspect=1.6)
g.set_axis_labels('Day of the Week', 'Temperature (°C)')
g.figure.suptitle('Weekly Temperature Trend (figure-level: relplot)', y=1.02)
plt.show()
```

**Difference observed:** `sns.lineplot()` is axes-level: it draws onto a single Matplotlib `Axes` (here `ax`), so you create the figure yourself and control size/titles with Matplotlib. `sns.relplot()` is figure-level: it creates its own figure (a `FacetGrid`) and returns it, sets size via `height`/`aspect`, and uses grid methods like `g.set_axis_labels()`. It can also split into multiple subplots with `col=` or `row=`.

## Task 4: Built-in themes

```python
for theme in ['darkgrid', 'whitegrid', 'ticks']:
    sns.set_theme(style=theme)
    plt.figure(figsize=(8, 5))
    sns.lineplot(x=days, y=temps, marker='o')
    plt.xlabel('Day of the Week')
    plt.ylabel('Temperature (°C)')
    plt.title(f'Weekly Temperature Trend ({theme})')
    plt.savefig(f'temperature_{theme}.png', dpi=150, bbox_inches='tight')
    plt.show()
```

**Comparison:**

- **darkgrid** (default): light-grey background with white grid lines on both axes.
- **whitegrid**: white background with light-grey grid lines.
- **ticks**: white background, no grid, with tick marks on the axes.

Three images are saved: `temperature_darkgrid.png`, `temperature_whitegrid.png`, `temperature_ticks.png`.
