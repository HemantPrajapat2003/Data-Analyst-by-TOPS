```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style='whitegrid')
rng = np.random.default_rng(42)
```

```python
n = 200
swiggy = pd.DataFrame({
    'order_id': range(1, n + 1),
    'payment_method': rng.choice(['UPI', 'Card', 'Cash', 'Wallet'], size=n, p=[0.45, 0.25, 0.2, 0.1]),
})

plt.figure(figsize=(7, 5))
sns.countplot(data=swiggy, x='payment_method', order=swiggy['payment_method'].value_counts().index)
plt.xlabel('Payment Method')
plt.ylabel('Number of Orders')
plt.title('Swiggy Orders by Payment Method')
plt.show()
```

```python
m = 300
zomato = pd.DataFrame({
    'food_category': rng.choice(['Pizza', 'Burger', 'Biryani', 'South Indian'], size=m),
    'orders': rng.integers(1, 6, size=m),
})
base = {'Pizza': 32, 'Burger': 28, 'Biryani': 40, 'South Indian': 25}
zomato['delivery_time'] = zomato['food_category'].map(base) + rng.normal(0, 5, size=m).round(1)

plt.figure(figsize=(7, 5))
sns.barplot(data=zomato, x='food_category', y='delivery_time', estimator='mean')
plt.xlabel('Food Category')
plt.ylabel('Average Delivery Time (minutes)')
plt.title('Average Delivery Time by Food Category')
plt.show()
```

```python
plt.figure(figsize=(7, 5))
sns.barplot(data=zomato, x='food_category', y='orders', estimator='sum', errorbar=None)
plt.xlabel('Food Category')
plt.ylabel('Total Number of Orders')
plt.title('Total Orders by Food Category')
plt.show()
```

```python
k = 250
bms = pd.DataFrame({
    'genre': rng.choice(['Action', 'Comedy', 'Drama', 'Thriller', 'Romance', 'Horror'], size=k),
    'tickets_sold': rng.integers(20, 200, size=k),
})

plt.figure(figsize=(8, 5))
sns.countplot(data=bms, x='genre', order=bms['genre'].value_counts().index)
plt.xlabel('Movie Genre')
plt.ylabel('Number of Shows')
plt.title('BookMyShow: Shows by Genre')
plt.show()

# Barplot with confidence interval removed (ci=None in old Seaborn, errorbar=None in new)
plt.figure(figsize=(8, 5))
sns.barplot(data=bms, x='genre', y='tickets_sold', estimator='mean', errorbar=None)
plt.xlabel('Movie Genre')
plt.ylabel('Average Tickets Sold')
plt.title('BookMyShow: Average Tickets Sold by Genre (no CI)')
plt.show()
```
