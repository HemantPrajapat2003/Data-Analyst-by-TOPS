```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style='whitegrid')
rng = np.random.default_rng(42)
```

Task 1, scatterplot of followers vs posts (20 users)

```python
insta = pd.DataFrame({
    'followers': rng.integers(200, 20000, 20),
    'posts': rng.integers(5, 400, 20),
})

plt.figure(figsize=(7, 5))
sns.scatterplot(data=insta, x='posts', y='followers')
plt.xlabel('Number of Posts')
plt.ylabel('Number of Followers')
plt.title('Instagram: Followers vs Posts (20 Users)')
plt.show()
```

Task 2, lineplot of daily Zomato orders (30 days)

```python
orders = rng.integers(80, 250, 30)
zomato = pd.DataFrame({'day': range(1, 31), 'orders': orders})

plt.figure(figsize=(10, 5))
sns.lineplot(data=zomato, x='day', y='orders', marker='o')
plt.xlabel('Day of the Month')
plt.ylabel('Number of Orders')
plt.title('Zomato Daily Orders Over a Month')
plt.show()
```

Task 3, relplot of rating vs reviews by genre (15 movies)

```python
movies = pd.DataFrame({
    'movie': [f'Movie {i}' for i in range(1, 16)],
    'rating': rng.uniform(5, 9.5, 15).round(1),
    'reviews': rng.integers(500, 20000, 15),
    'genre': ['Action'] * 5 + ['Comedy'] * 5 + ['Drama'] * 5,
})

g = sns.relplot(data=movies, x='rating', y='reviews', hue='genre', height=5, aspect=1.4)
g.set_axis_labels('Movie Rating', 'Number of Reviews')
g.figure.suptitle('Movie Ratings vs Reviews by Genre', y=1.02)
plt.show()
```

Task 4, regplot of YouTube hours vs exam scores (25 students)

```python
hours = rng.uniform(0, 6, 25).round(1)
scores = (88 - hours * 6 + rng.normal(0, 6, 25)).clip(0, 100).round(0)
students = pd.DataFrame({'youtube_hours': hours, 'exam_score': scores})

plt.figure(figsize=(7, 5))
sns.regplot(data=students, x='youtube_hours', y='exam_score')
plt.xlabel('Hours Spent on YouTube (per day)')
plt.ylabel('Exam Score')
plt.title('YouTube Hours vs Exam Scores')
plt.show()

print('Correlation:', round(students['youtube_hours'].corr(students['exam_score']), 2))
```
