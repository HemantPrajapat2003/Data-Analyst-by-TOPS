```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style='darkgrid')
```

Task 2, hue by sex

Task 1, basic pairplot

```python
tips = sns.load_dataset('tips')
print(tips.head())
print(tips.select_dtypes('number').columns.tolist())

sns.pairplot(tips)
plt.show()
```

Task 2, hue by sex

```python
sns.pairplot(tips, hue='sex')
plt.show()
```

Task 3, hue by smoker with KDE on the diagonal

```python
sns.pairplot(tips, hue='smoker', diag_kind='kde')
plt.show()
```

Task 4, two columns only with hue by time


```python
sns.pairplot(tips, vars=['total_bill', 'tip'], hue='time')
plt.show()
```
