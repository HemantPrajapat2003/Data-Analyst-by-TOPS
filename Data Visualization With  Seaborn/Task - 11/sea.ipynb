```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme()
rng = np.random.default_rng(42)
```

Task 1, barplot saved as a 300 DPI PNG

```python
orders = pd.DataFrame({
    'category': ['Pizza', 'Burger', 'Biryani', 'Momos', 'Ice Cream'],
    'orders': [320, 280, 410, 190, 150],
})

plt.figure(figsize=(8, 5))
sns.barplot(data=orders, x='category', y='orders', hue='category', legend=False)
plt.xlabel('Food Category')
plt.ylabel('Number of Orders')
plt.title('Orders per Food Category')
plt.savefig('orders_300dpi.png', dpi=300, bbox_inches='tight')
plt.show()
```

Task 2, lineplot saved as an SVG

```python
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
steps = [6200, 8100, 7400, 9000, 10200, 12500, 7800]

plt.figure(figsize=(8, 5))
sns.lineplot(x=days, y=steps, marker='o')
plt.xlabel('Day')
plt.ylabel('Steps')
plt.title('Daily Step Count - One Week')
plt.savefig('steps_week.svg', format='svg', bbox_inches='tight')
plt.show()
```

Task 3, darkgrid with a custom palette at 1200 DPI

```python
sns.set_theme(style='darkgrid', palette='viridis')

plt.figure(figsize=(6, 4))
sns.barplot(data=orders, x='category', y='orders', hue='category', palette='viridis', legend=False)
plt.xlabel('Food Category')
plt.ylabel('Number of Orders')
plt.title('Orders per Food Category (darkgrid, viridis)')
plt.savefig('orders_1200dpi.png', dpi=1200, bbox_inches='tight')
plt.show()
```

Task 4, theme template and pie chart of top 5 apps

```python
def apply_presentation_theme():
    """Apply one consistent look: style, font, context and palette."""
    sns.set_theme(
        style='whitegrid',
        context='talk',
        palette='deep',
        font='DejaVu Sans',
        font_scale=1.0,
    )
    plt.rcParams['axes.titleweight'] = 'bold'

apply_presentation_theme()

apps = ['WhatsApp', 'Instagram', 'YouTube', 'Chrome', 'Spotify']
hours = [3.5, 2.5, 2.0, 1.5, 1.0]   # replace with your own screen-time numbers

plt.figure(figsize=(7, 7))
plt.pie(hours, labels=apps, autopct='%1.1f%%', startangle=90, colors=sns.color_palette('deep', 5))
plt.title('My Top 5 Most-Used Apps')
plt.savefig('top_apps_pie.png', dpi=300, bbox_inches='tight')
plt.show()
```

Task 5, heatmap exported as an SVG

```python
sns.set_theme()
data = pd.DataFrame(rng.integers(50, 300, (5, 7)),
                    index=['Pizza', 'Burger', 'Biryani', 'Momos', 'Ice Cream'],
                    columns=['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun'])

plt.figure(figsize=(8, 5))
sns.heatmap(data, annot=True, fmt='d', cmap='YlGnBu')
plt.title('Orders by Food Category and Day')
plt.savefig('orders_heatmap.svg', format='svg', dpi=1200, bbox_inches='tight')
plt.show()
```
